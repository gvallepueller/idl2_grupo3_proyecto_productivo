1. Calcular la suma y promedio de los números pares e impares por separado, de 1 hasta N.

```python
# Resolución de Ejercicio 1:
#Creamos una función en donde pedimos como entrada "n", es decir el límite de nuestra lista de números
def cacular_suma_promedio(n:int):
    # Inciamos las variables que comienzan en 0 dentro de la función
    suma_pares: int = 0
    cantidad_pares: int = 0
    suma_impares: int = 0
    cantidad_impares: int = 0

    # Creamos un bucle que crea nuestra lista contando de uno hasta "n"
    for numero in range(1, n + 1): #Se usa "n+1" para incluir el número de entrada

        #Evaluamos si modulo es 0 entonces es par
        if numero % 2 == 0:
            suma_pares += numero      # Acumulamos el valor sumandolo a "suma pares"
            cantidad_pares += 1       # Contamos la cantidad de pares sumando 1 a "cantidad de pares"

        #Si el número es impar se hace lo mismo pero sumamos a la suma impares.
        else:
            suma_impares += numero
            cantidad_impares += 1

    # Para no dividir entre 0 comprobamos que la cantidad sea mayor que 0
    if cantidad_pares > 0:
        promedio_pares = suma_pares / cantidad_pares #Si lo es hacemos el promedio
    else:
        promedio_pares = 0
    #Igualmente con los impares.
    if cantidad_impares > 0:
        promedio_impares = suma_impares / cantidad_impares
    else:
        promedio_impares = 0


    #Imprimimos resultados:
    print(f"\nLos resultados de Pares son: \n   Suma: {suma_pares} \n   Promedio: {promedio_pares}")
    print(f"\nLos resultados de Impares son: \n   Suma: {suma_impares} \n   Promedio: {promedio_impares}")

    return "\nPrograma terminado."

# Pedimos al usuario el número inicial:
input_usuario = int(input("Ingrese el número inicial: "))
resultado = cacular_suma_promedio(input_usuario)
print(resultado)
```

2. Realizar un algoritmo que permita pedir 10 números y determine e imprima cuantos son pares, impares, positivos, neutros y negativos.

>> **IMPORTANTE:** *Un número es par si al dividirse entre 2 da 0. Por tal motivo 0 es teóricamente par. Pero, para el presente ejercicio, 0 no se está considerando ni par, ni impar.*

```python
# Resolución de Ejercicio 2:
# Creamos una función; pero, no pedimos datos de entrada porque los pediremos dentro de la variable.
def analizar_numeros():
    # Iniciamos las variables desde 0 dentro de la función
    cantidad_pares: int = 0
    cantidad_impares: int = 0
    cantidad_positivos: int = 0
    cantidad_negativos: int = 0
    cantidad_neutros: int = 0

    # Vamos a crear una lista donde guardermos todos los datos
    lista_numeros: list[int] = []


    # Creamos una variable que no servirá para contar los números que ingresen
    numero: int = 0

    # Creamos un bucle que se repite 10 veces desde el indice 0 al 9:
    for i in range(10):
        # Pedimos al usuario el número
        # Indicamos cuantos números faltan usando "i + 1" es decir para que no empiece en indice 0 se le suma 1.
        numero = int(input(f"Ingrese el número {i + 1} de 10: "))

        # Agregamos el número al final de la lista usando el método .append()
        lista_numeros.append(numero)

        # Evaluamos si el modulo es 0, si al dividir entre 2 es 0 entonces es par.
        if numero % 2 == 0 and numero != 0:
            cantidad_pares += 1 # Sumamos 1 al contador de pares
        # Si el número no es par entonces es impar
        elif numero % 2 != 0 and numero != 0:
            cantidad_impares += 1 # Sumamos 1 al contador de impares

        # Creamos una condicional para ver si es negativo, positivo o neutro (0)
        if numero > 0:
            cantidad_positivos += 1   # Si es mayor a 0 se suma a positivos
        elif numero < 0:
            cantidad_negativos += 1   # Si no es mayor a 0 evaluamos si es menor a 0, si lo es se suma a negativos
        else:
            cantidad_neutros += 1     # Si no es ninguno entonces es neutro, es decir 0


        """ Otra forma de comprobar si es 0, par o impar
        if numero == 0:
            cantidad_neutros += 1
        else:
            if numero > 0:
                cantidad_positivos += 1
            else:
                cantidad_negativos += 1
            if numero % 2 == 0:
                cantidad_pares += 1
            else:
                cantidad_impares += 1
        """


    # Imprimimos resultados:
    print("\nLos números ingresados son: \n  ", lista_numeros)
    print("\nResultados del análisis:")
    print(f"  La cantidad de Pares es: {cantidad_pares}")
    print(f"  La cantidad de Impares es: {cantidad_impares}")
    print(f"  La cantidad de Positivos es: {cantidad_positivos}")
    print(f"  La cantidad de Negativos es: {cantidad_negativos}")
    print(f"  La cantidad de Neutros es: {cantidad_neutros}")

    return "\nPrograma terminado."

# Ejecutamos la función directamente ya que dentro de esta se pide el input
resultado = analizar_numeros()
print(resultado)
```

3. Hacer un programa que permita ingresar un conjunto de N números y un numero más X mostrar como resultado:

    - Porcentaje de números menores que X.
    - Porcentaje de números mayores de X.
    - Porcentaje de números iguales de X.
    - Porcentaje de números comprendidos entre 1 y X.
    - Porcentaje de números que no cumple ninguna condición

```python
# Resolución de Ejercicio 3:
# Creamos una función que tendrá dentro las entradas:
def analizar_conjunto():

    # Inicializamos los contadores en 0
    menores_x: int = 0
    mayores_x: int = 0
    iguales_x: int = 0
    entre_1_x: int = 0
    ninguna: int = 0

    # Creamos una lista para almacenar el historial de números
    lista_numeros: list[int] = []

    # Cremos la variables "N" y "X"
    n: int = 0
    x: int = 0

    # Creamos una variable que no servirá para contar los números que ingresen
    numero: int = 0


    #Solicitamos "N" o el número de elementos en el conjunto
    n = int(input("Ingrese la cantidad de números que desea evaluar (N): "))

    #Si N es 0 o negativo, no podemos calcular porcentajes más tarde.
    #Verificamos si n es mayor a 0, si no lo es retornamos un error
    if n <= 0:
        return "ERROT: Ingrese un número mayor a 0."

    #Pedimos el números "X" que será el número de análisis
    x = int(input("Ingrese el número de análisis (X): "))

    # Hacemos un bucle usando la cantidad de números N
    for i in range(n):
        # Pedimos el número usando i + 1 para que no se vea el indice 0
        numero = int(input(f"Ingrese el número {i + 1} de {n}: "))
        lista_numeros.append(numero) # añadimos ese número a la lista


        # Verificamos a qué grupo pertenece con condicionales
        if numero < x:
            menores_x += 1 #Si es menor se suma a cantidad menores
        elif numero > x:
            mayores_x += 1 #Si es mayor se suma a cantidad mayor
        elif numero == x:
            iguales_x += 1 #Si es igual se suma a cantidad igual
        else:
            ninguna += 1 #Si no es ninguno se suma a ninguna

        #Si el número esta entre 1 y X lo sumamos a entre_1_x
        if 1 < numero < x:
            entre_1_x += 1

    # Usamos la formula: (cantidad / total) * 100 para calcular porcentajes:
    # Retornamos float porque saldrán decimales.
    porc_menores: float = (menores_x / n) * 100
    porc_mayores: float = (mayores_x / n) * 100
    porc_iguales: float = (iguales_x / n) * 100
    porc_entre: float = (entre_1_x / n) * 100
    porc_ninguna: float = (ninguna / n) * 100

    # Impresión de resultados:
    print("\nLos números ingresados son: \n  ", lista_numeros)
    print(f"\nResultados del análisis respecto a X ({x}):")
    print(f"  Menores que X: {porc_menores:.2f}%, ({menores_x })") #usamos el formato ".2f" para solo mostrar 2 decimales
    print(f"  Mayores que X: {porc_mayores:.2f}%,  ({mayores_x })")
    print(f"  Iguales a X: {porc_iguales:.2f}%,  ({iguales_x })")
    print(f"  Comprendidos entre 1 y X: {porc_entre:.2f}%,  ({entre_1_x })")
    print(f"  Sin condición: {porc_ninguna:.2f}%,  ({ninguna})")

    return "\nPrograma terminado."

# Ejecución de la función
resultado = analizar_conjunto()
print(resultado)
```

4. De una lista de N números determinar simultáneamente el máximo y mínimo número.


```python
# Resolución de Ejercicio 4:
#Creamos la función sin datos de inicio ya que los pedirá:
def determinar_max_min():

    # Creamos dos variables  de maximo y minimo
    maximo: int = 0
    minimo: int = 0

    # Creamos una variable que no servirá para contar los números que ingresen
    numero: int = 0

    #Solicitamos "N" o el número de elementos en el conjunto
    n: int = int(input("Ingrese la cantidad de números que desea evaluar (N): "))

    #Si N es 0 o negativo no habrá cantidad de números a contar.
    #Verificamos si n es mayor a 0, si no lo es retornamos un error
    if n <= 0:
        return "ERROT: Ingrese un número mayor a 0."

    lista_numeros: list[int] = [] # Creamos una lista para almacenar el historial de números

    # Hacemos un bucle usando la cantidad de números N
    for i in range(n):
        #Pedimos al uusario un números y le indicamos cuantos faltan usando i+1
        numero = int(input(f"Ingrese el número {i + 1} de {n}: "))
        lista_numeros.append(numero) # añadimos ese número a la lista

    #Aplicamos "max" y "min" a la lista para poder saber cual número es el maximo y el minimo
    maximo = max(lista_numeros)
    minimo = min(lista_numeros)

    #Imprimimos los resultados
    print("\nLos números ingresados son: \n  ", lista_numeros)
    print("\nResultados del análisis:")
    print(f"  El número máximo es: {maximo}")
    print(f"  El número mínimo es: {minimo}")

    return "\nPrograma terminado."

resultado = determinar_max_min()
print(resultado)
```

5. Realizar un algoritmo para determinar cuánto ahorrará una persona en un año, si al final de cada mes deposita cantidades variables de dinero; se quiere saber cuánto lleva ahorrado cada vez que haga un deposito en cada mes y luego mostrar el ahorro final de 12 meses.

```python
# Resolución de Ejercicio 5:

def calcular_ahorro():
    # Creamos una tupla con los meses ya que no queremos cambiar este arreglo.
    # Los meses serán inmutables
    meses: tuple = ("Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio","Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre")
    #También el deposito del mes
    deposito: str = "" #sirve para que se escriba la cantidad del mes o  "salir" para terminar el programa

    # Creamos una variable del total acumulado en floar ya que puede contener decimales.
    ahorro_acumulado: float = 0.0

    # Creamos una lista vacía que servirá como base de datos de nuestro historial.
    historial_depositos: list[dict] = []

    # Creamos un diccionario que guardará el mes, el deposito y el acumulado.
    # La idea es que este diccionario esté dentro de la lista "historial_depositos"
    registro_mensual: dict = {}

    print("--- Sistema de Ahorro Anual ---")

    for mes in meses:
        # Pedimos la cantidad a depositar
        # Si el cliente desea parar le decimos que escriba salir
        deposito = (input(f"Ingrese la cantidad a depositar en {mes}\nEscriba salir para terminar:\n"))

        #Usamos strip y lower para que todo lo que escriba el usuario sea sin separación y en minúsculas
        if deposito.strip().lower() == "salir":
            break # Si escribió salir se rompe el bucle
        else:
            deposito = float(deposito) #Sino se transforman los números a floats

        #Sumamos al ahorro acumulado el depósito
        ahorro_acumulado += deposito

        #Mostramos el ahorro acumulado hasta ese mes con un formado de 2 decimales con .2f
        print(f"  -> Ahorro total acumulado hasta {mes}: ${ahorro_acumulado:.2f}\n")

        # Añadimos al registro mensual el mes, el deposito y el ahorro
        registro_mensual: dict = {
            "Mes": mes,
            "Deposito": deposito,
            "Acumulado": ahorro_acumulado
        }
        historial_depositos.append(registro_mensual) #luego añadimos ese diccionario a la lista de historial de de positos.



    # Con un bucle mostramos lo ahorrado hasta fin de año:
    print("--- Ahorro Anual ---")
    for registro in historial_depositos:
        # Con las claves del diccinario revisamos mes por mes cada deposito y el acumulado
        print(f"{registro['Mes']}: Depositó ${registro['Deposito']:.2f} | Total acumulado: ${registro['Acumulado']:.2f}")

    # Mostramos el total ahorrado:
    print(f"\nEl ahorro después de 12 meses es: ${ahorro_acumulado:.2f}")

    return "\nPrograma terminado."

# Ejecución de la función
resultado = calcular_ahorro()
print(resultado)
```

6. Hacer un programa que muestre un cronometro (hh:mm:ss), indicando las horas, minutos y segundos(Considerar que el tiempo no será exacto al de un cronometro real solo será referencial)

>> **IMPORTANTE:** *Para crear una *SIMULACIÓN* de cronómetro mostrarémos un "cronómetro" que mostrará el tiempo de 30 en 30 segundos. El cronómetro se detendrá a las 2 horas, e importaremos "time" para poder mostrar el cronómetro cada 0.3 segundos*

```python
# Resolución de Ejercicio 6:
import time #Importamos time para poder detener el proceso cada cierto tiempo y mostrar lentamente el cronómetro
def simular_cronometro():

    print("--- Iniciando simulación de cronómetro (Alta velocidad) ---")

    #Hacemos un primer bucle, este bucle contará las horas, para la simulación contará solo 2 horas
    for hora in range(2):

        #Dentro del bucle se contarán los mnutos, esto será en rango de 60 por 60 minutos
        #Contará del 0 al 59
        for minuto in range(60):

            # Luego haremos el bucle de segundos.
            # En rango de 60 porque contará del 0 al 59 segundos.
            # Con paso de 30 para que muestre cada 30 segundos.
            for segundo in range(0, 60, 30):

                #Usamos el formato ".02d" esto hace que se fuercen números de 2 digitos con un 0 a la izquierda.
                print(f"{hora:02d}:{minuto:02d}:{segundo:02d}")
                time.sleep(0.3)

    return "\nSimulación finalizada."

# Ejecución de la función
resultado = simular_cronometro()
print(resultado)
```

7. Crea una aplicación que pida dos números y calcule su factorial de cada
número, pero el programa también tiene que indicar hasta que valor
serian. Ejemplo:
    - 7! > 1x2x3x4x5x6x7
    - 10! > 1x2x3x4x5x6x7x8x9x10
      - Son iguales hasta el múltiplo 7

```python
# Resolución de Ejercicio 7:
def comparar_factorial(): #creamos una función que pedirá los datos
    # Pedimos los dos números
    numero_1: int = int(input("Ingrese el primer número: "))
    numero_2: int = int(input("Ingrese el segundo número: "))

    # Se hará un concatenación de cadenas de texto para mostrar la multuplicación del factorial.
    # Por esta razón creados 2 variables de string vacías.
    text_fact_1: str = ""
    text_fact_2: str = ""

    #Creamos las variables para multiplicar, inician en 1 porque si colocamos 0 se multiplicará por 0
    resultado_fact_1: int = 1
    resultado_fact_2: int = 1
    limite_igualdad: int = 0

    # Creamos un bucle para poder calcular el factorial
    for i in range(1, numero_1 + 1):
        # Luego en al resultado lo multiplicamos por el número actual.
        resultado_fact_1 *= i

        # Añadimos al texto 1 el número actual qeu se está multiplicando
        text_fact_1 += str(i)

        # Si el número es menor al número final del factorial colocamos despues el simbolo "x"
        if i < numero_1:
            text_fact_1 += "x"

    # Repetimos el mismo bucle para el segundo factorial
    for i in range(1, numero_2 + 1):
        resultado_fact_2 *= i
        text_fact_2 += str(i)
        if i < numero_2:
            text_fact_2  += "x"

    #Para indicar hasta que valor serian iguales solo debemos buscar el factorial menor
    #Ya que el factorial menor es el multiplo común final.
    #Usamos una condicional para saber cual es el menor:
    if numero_1 < numero_2:
        limite_igualdad = numero_1
    else:
        limite_igualdad = numero_2

    # Mostramos la comparación
    print("\nComparación de factoriales:")
    print(f"\n  - {numero_1}! > {text_fact_1}\n    (Total: {resultado_fact_1})")
    print(f"\n  - {numero_2}! > {text_fact_2}\n    (Total: {resultado_fact_2})")
    print(f"\n   - Son iguales hasta el múltiplo {limite_igualdad}!")

    return "\nPrograma terminado."

# Ejecución de la función
resultado = comparar_factorial()
print(resultado)
```

8. A una fiesta asistieron personas de diferentes edades y sexos. Construir
un algoritmo donde se ingrese los datos de los N asistentes. Calcular:
    - Cuantas personas asistieron a la fiesta
    - Cuantos hombres y cuantas mujeres
    - Promedio de edades por sexo
    - La edad de la persona de menor y mayor edad que asistió.

Nota: El sistema debe permitir al menos el ingreso los datos de una
persona y solicitar si se desasea registrar a otra.

```python
# Resolución de Ejercicio 8:
def analizar_asistentes_fiesta():
    # Declaramos las variables de control
    total_asistentes: int = 0
    total_hombres: int = 0
    total_mujeres: int = 0

    # Declaramos las variables de acumulación y promedio
    suma_edad_hombres: int = 0
    suma_edad_mujeres: int = 0

    # Declaramos las variables de edad máxima y mínima.
    # En este caso se usa -inf (menos infinito) para que cualquier número ingresado en "edad máxima" siempre sea mayor al valor inicial.
    # En este caso se usa inf (más infinito) para que cualquier número ingresado en "edad mínima" siempre sea menor al valor inicial.
    edad_maxima: float = float('-inf')
    edad_minima: float = float('inf')

    # Declaramos las variables de promedio
    promedio_mujeres: float = 0.0
    promedio_hombres: float = 0.0

    print("--- Registro de Asistencia ---")

    #Se inicia un bucle para ir anotando uno por uno a los asistentes.
    while True:
        # Solicitamos la edad
        edad: int = int(input("\nIngrese la edad de la persona: "))

        # Para saber el sexo indicamos que escriba H o M y con .upper nos aseguramos que al respues siempre sea mayúscula.
        sexo: str = input("Ingrese el sexo de la persona (H para hombre o M para mujer): ").upper()

        # Sumamos 1 a los asistentes
        total_asistentes += 1

        # Con una condicional si es hombres sumamos 1 a los asistentes hombres y sumamos la edad para hacer luego el promedio.
        if sexo == 'H':
            total_hombres += 1
            suma_edad_hombres += edad
        # Hacemos lo mismo con las mujeres.
        elif sexo == 'M':
            total_mujeres += 1
            suma_edad_mujeres += edad

        #Con otra condicional verificamos si la edad actual es mayor al edad máxima se actualiza
        if edad > edad_maxima:
            edad_maxima = edad #como edad maxima inicia en "-inf" la primera edad siempre será la maxima
        #De la misma manera con la edad minima
        elif edad < edad_minima:
            edad_minima = edad #Como la edad minima inicia en "inf" la primera edad tambien será la edad mínima.

        #Ahora preguntamos al cliente si desea añadir otra persona a la lista
        #Usamos un string con .lower para que su respuesta S o N siempre sea minuscula
        respuesta: str = input("\n¿Desea registrar a otra persona? (s/n): ").lower()

        if respuesta != 's': #Si la respuesta no es s
            break #entonces se rompe el bucle

    #Ahora creamos el promedio, si el total de hombres es mayor a 0
    #Se divide la suma de edades entre el total de hombres
    if total_hombres > 0:
        promedio_hombres = suma_edad_hombres / total_hombres

    #Hacemos lo mismo para las mujeres.
    if total_mujeres > 0:
        promedio_mujeres = suma_edad_mujeres / total_mujeres

    #Por ultimo imprimimos los datos
    print("\n--- Resultados del Análisis de Asistentes---")
    print(f"\n  Total de personas asistentes: {total_asistentes}")
    print(f"\n  Total de hombres: {total_hombres}")
    print(f"\n  Total de mujeres: {total_mujeres}")
    print(f"\n  Promedio de edad en hombres: {promedio_hombres:.2f} años") #usamos .2f porque el promedio puede ser decimal.
    print(f"\n  Promedio de edad en mujeres: {promedio_mujeres:.2f} años")
    print(f"\n  Edad mínima registrada: {edad_minima} años")
    print(f"\n  Edad máxima registrada: {edad_maxima} años")

    return "\nPrograma terminado."

# Ejecución de la función
resultado = analizar_asistentes_fiesta()
print(resultado)
```

#
9. Calcular el producto utilizando arrays: 100 x 98 x 96 x 94 x . . . x 1 en
este orden.

```python
# Resolución de Ejercicio 9:
def calcular_semifactorial():
    # Creamos una lista para guardar toda la lista de números de 100 a 1
    lista_numeros: list[int] = []

    #Ahora creamos una variable donde añadiremos la multiplicación de todos los números
    producto_total: int = 1

    # Utilizamos un bucle flor declarando el incio en 100, el final excluyente en 0 (no cuenta el 0 sino hasta el 1).
    # Cómo queremos que descienda de 2 en dos el paso será de -2
    for numero in range(100, 0, -2):
        lista_numeros.append(numero) #con append añadimos el número a la lista de secuencia.

    #Imprimimos la lista de los números
    print("Lista de números en orden descendiente del 100 al 1, de 2 en 2:")
    print(lista_numeros)

    #Con un bucle recorremos las lista del 100 al 1
    for valor in lista_numeros:
        #vamos multiplicando el valor de cada número por el producto total y actualizando la cantidad.
        producto_total *= valor

    # Por ultimo imprimimos en pantalla los resultados:
    print("\nResultados del análisis:")
    print(f"  Cantidad números en al lista:\n    {len(lista_numeros)} elementos.")
    print(f"  El producto total del semifactorial o doble factorial 100!! es:\n    {producto_total}")

    return "\nPrograma terminado."

# Ejecutamos
resultado = calcular_semifactorial()
print(resultado)
```

    8. La cantidad de alumnos aprobados

    10. El porcentaje de alumnos aprobados
    11. El porcentaje de alumnos desaprobados

```python
# Resolución de Ejercicio 10:
def analizar_calificaciones():
    #Iniciamos una variable para controlar la cantidad de estudiantes y las notas
    n: int = 0
    nota: int = 0

    # Ahora inicializamos las variables de notas maxima y minima
    nota_maxima: float = float('-inf') #-inf para que la primera nota siempre sea la nota mayor
    nota_minima: float = float('inf')#inf para que la primera nota minima siempre sea la menor

    #Inicializamos el contador de alumnos con notas mínimas y maximas:
    alumnos_maxima: int = 0
    alumnos_minima: int = 0

    #Creamos los contadores de las notas que solicita el ejercicio.
    suma_notas: int = 0
    notas_pares: int = 0
    notas_impares: int = 0
    aprobados: int = 0
    desaprobados: int = 0

    #Creamos las variables aritmétricas
    promedio_salon: float = 0.0
    porce_aprobados: float = 0.0
    porce_desaprobados: float = 0.0

    #Creamos una lista donde estarán todas la noches:
    lista_notas: list[int] = []

    #Con un bucle while nos aseguramos que el número sea entre 5 y 20:
    while True:
        n = int(input("Ingrese la cantidad de estudiantes. Debe ser entre 5 y 20): "))
        # Evaluamos el rango exigido por el problema
        if 5 <= n <= 20:
            break

        else:
            print("  [Error de Entrada] El valor debe estar estrictamente comprendido entre 5 y 20.\n")

    print("\n--- Ingrese las Calificaciones ---")

    #Ahora usamos un bucle for donde el rango de iteración será la cantidad de alumnos:
    for i in range(n):
        while True: #dentro del bucle for usamos un while para verificar que la nota sea entre 0 y 20
            # Pedimos la nota y usamos i + 1 para no mostrar un indice 0
            nota = int(input(f"Ingrese la calificación del estudiante {i + 1} (0-20): "))

            #Con una condicional verificamos que la nota no sea menor a 0 ni mayor a 20
            if 0 <= nota <= 20:
                lista_notas.append(nota) #Si la nota es correcta se añade a la lista de de notas
                break #Rompemos el ciclo while para seguir con la siguiente nota.
            else:
                print("  [ERROR] Ingrese un número entre 0 y 20.\n") #Sino lanzamos una advertencia.


    #Ahora vamos a recorrer las la lista de notas para poder analizar cada una con un for
    for nota in lista_notas:
        suma_notas += nota #sumamos las notas actualizando las variables para poder promediarlas luego.

        #Comprobamos si es par con el modulo o resto:
        if nota % 2 == 0:
            notas_pares += 1 #Si la división entre 2 da 0 se suma a pares 1.
        else:
            notas_impares += 1 #si la división entre 2 no da 0 se suma a impares 1

        #Creamos una condición para contar los aprobados
        if nota >= 13:
            aprobados += 1 #si es 13 o más se suma 1 a los aprobados.
        else:
            desaprobados += 1 #Sino se suma 1 a los desarpobados.

        #Ahora creamos una condicional para validar las notas máximas
        if nota > nota_maxima: #Si la nota actual del analisis es mayor a la nota maxima
            nota_maxima = nota # Se actualiza la nota maxima
            alumnos_maxima = 1   # Reiniciamos el contador porque hay un nuevo récord
        elif nota == nota_maxima: #Si la nota actual es igual a la maxima
            alumnos_maxima += 1  # Sumamos 1 si alguien igualó el récord actual

        #Ahora hacemos lo mismo para las notas mínimas
        if nota < nota_minima:
            nota_minima = nota
            alumnos_minima = 1
        elif nota == nota_minima:
            alumnos_minima += 1

    #Por ultimo creamos los calculos con float pues pueden tener decimales
    promedio_salon = suma_notas / n
    porce_aprobados = (aprobados / n) * 100
    porce_desaprobados = (desaprobados / n) * 100

    # 6. Salida de Datos (Reporte Estadístico Formateado)
    print("\n--- Reporte Final del Salón ---")
    print(f"  Calificación Mayor: {nota_maxima} -- Obtenida por {alumnos_maxima} alumno/s)")
    print(f"  Calificación Menor: {nota_minima} -- Obtenida por {alumnos_minima} alumno/s)")
    print(f"  Promedio general del salón: {promedio_salon:.2f}") #usamos .2f para 2 decimales
    print(f"  Cantidad de notas pares: {notas_pares}")
    print(f"  Cantidad de notas impares: {notas_impares}")
    print(f"  Alumnos Aprobados (>=13): {aprobados} ({porce_aprobados:.2f}%)")
    print(f"  Alumnos Desaprobados (<13): {desaprobados} ({porce_desaprobados:.2f}%)")

# Ejecución de la función
analizar_calificaciones()
```
