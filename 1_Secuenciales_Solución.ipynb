1.  Determina la velocidad final de un auto si su velocidad inicial es 0 km/h y su aceleración es 0.8 m/s2; y el tiempo transcurrido es de 30 segundos.

>> IMPORTANTE: Se usará la formula:
>> - Velocidad Final = Velocidad Inicial + ( Aceleración x Tiempo )

```python
# Resolución de Ejercicio 1:
#Definimos las variables como flotantes ya que pueden tener decimales
KM = 1000 #contante de kilometros en metros
HORA = 3600 #constante de hora en segundos

velo_ini_kmh: float = 0.0
velo_ini_ms: float = 0.0
velo_final_ms: float = 0.0
ace_ms2: float = 0.8
tiempo_s: float = 30.0


# Pedimos la velocidad
velo_ini_kmh: float = float(input("Ingrese la velicidad inicial en kilómetros: "))

#Transformamos la velocidad en incial de Km a metros por segundo simplificando dividiendo la hora en segundos y el km en metros.
velo_ini_ms = velo_ini_kmh / (HORA/KM)

#Aplicamos la formula:
#Velocidad final = velocidad inicial + (acelaración * tiempo)
velo_final_ms = velo_ini_ms + (ace_ms2 * tiempo_s)

#Ahora pasamos la velocidad final a KH multiplicando la hora en segundo entre los metros en segundos
velo_final_kmh = velo_final_ms * (HORA/KM)

# 5. Salida de Datos (Reporte Formateado)
print("--- Resultado: ---")
print(f"  Velocidad Inicial: {velo_ini_ms:.2f} m/s") #usamos .2f para dos decimales
print(f"  Aceleración: {ace_ms2:.2f} m/s²")
print(f"  Tiempo transcurrido: {tiempo_s:.2f} s")
print("-------------------------------------------")
print(f"  Velocidad Final en metros por segundo: {velo_final_ms:.2f} m/s")
print(f"  Velocidad Final en Kilómetros por hora: {velo_final_kmh:.2f} km/h")

print("\nCálculo terminado.")
```

3. Tres personas deciden invertir su dinero para fundar una empresa. Cada una de ellas invierte una cantidad distinta. Obtener el porcentaje que cada uno invierte con respecto a la cantidad total invertida.

```python
# Resolución de Ejercicio 3:

print("--- Inversiones porcentuales en la empresas: ---")

#Inicializamos la variables de la suma de las inversiones y los procentajes
suma_inversiones: float = 0.0
porce_inver_1: float = 0.0
porce_inver_2: float = 0.0
porce_inver_3: float = 0.0
inver_1: float = 0.0
inver_2: float = 0.0
inver_3: float = 0.0

#Solicitamos los montos de cada uno las inversiones como flotat ya que tienen decimales
inver_1 = float(input("Ingrese el monto de la primera inversión: "))
inver_2 = float(input("Ingrese el monto de la segunda inversión: "))
inver_3 = float(input("Ingrese el monto de la tercera inversión: "))

#Por ultimo creamos los caulculos, necesitamos saber cuanto vale el "100%"
#Por tal motivo sumamos el total de las inversiones:
suma_inversiones = (inver_1 + inver_2 + inver_3)

#Ahora varificamos el procentaje dividiendo el monto entre el total y multiplicandolo por 100
porce_inver_1 = (inver_1 / suma_inversiones) * 100
porce_inver_2 = (inver_2 / suma_inversiones) * 100
porce_inver_3 = (inver_3 / suma_inversiones) * 100

#Por ultimo imprimimos los resultados en pantalla
print("\n--- Cálculo de Participación Empresarial ---")
print(f"\nEl monto total invertido es de: {suma_inversiones}")
print(f"\nEl primer porcentaje es:  {porce_inver_1:.2f}%, corresponde al monto de: S/. {inver_1:.2f}")
print(f"El segundo porcentaje es: {porce_inver_2:.2f}%, corresponde al monto de: S/. {inver_2:.2f}")
print(f"El tercer porcentaje es:  {porce_inver_3:.2f}%, corresponde al monto de: S/. {inver_3:.2f}")
```

4. Un vendedor recibe su sueldo base más 25% extra por comisión de sus ventas; el vendedor desea saber cuánto dinero recibirá por concepto de comisiones por las tres ventas que realiza en el mes y el total que recibirá en el mes tomando en cuenta sueldo base y comisiones.


```python
# Resolución de Ejercicio 4:

print("--- Sistema de Cálculo de Sueldo y Comisiones ---")

# Inicializamos las variables de estado en 0.0 como buena práctica
PORCENTAJE_COMISION: float = 0.25
sueldo_base: float = 0.0
venta_uno: float = 0.0
venta_dos: float = 0.0
venta_tres: float = 0.0
total_ventas: float = 0.0
comision_venta_uno: float = 0.0
comision_venta_dos: float = 0.0
comision_venta_tres: float = 0.0
total_comisiones: float = 0.0
sueldo_total: float = 0.0

# Solicitamos los datos al usuario
# Se utiliza float ya que los montos monetarios incluyen decimales
sueldo_base = float(input("Ingrese su sueldo base: S/ "))
venta_uno = float(input("Ingrese el monto de la primera venta: S/ "))
venta_dos = float(input("Ingrese el monto de la segunda venta: S/ "))
venta_tres = float(input("Ingrese el monto de la tercera venta: S/ "))

#Primero  calculamos la suma total de las ventas realizadas en el mes
total_ventas = (venta_uno + venta_dos + venta_tres)

#Luego calculamos la cominsión individual por ventas
#La comisión individual se hace multiplicnado la venta por la comisión
comision_venta_uno = venta_uno * PORCENTAJE_COMISION
comision_venta_dos = venta_dos * PORCENTAJE_COMISION
comision_venta_tres = venta_tres * PORCENTAJE_COMISION

#Vamos a calcular la comision total:
total_comisiones = comision_venta_uno + comision_venta_dos + comision_venta_tres

# Finalmente, calculamos el sueldo neto sumando la base más las comisiones
sueldo_total = sueldo_base + total_comisiones

# Imprimimos los resultados en pantalla estructurados y con dos decimales
print("\n--- Reporte de Pagos Mensuales ---")
print(f"  Sueldo Base establecido: S/ {sueldo_base:.2f}")
print(f"  Venta 1:                 S/ {venta_uno:.2f} -- Corresponde a S/. {comision_venta_uno:.2f} de comisión")
print(f"  Venta 2:                 S/ {venta_dos:.2f} -- Corresponde a S/. {comision_venta_dos:.2f} de comisión")
print(f"  Venta 3:                 S/ {venta_tres:.2f} -- Corresponde a S/. {comision_venta_tres:.2f} de comisión")
print("-----------------------------------------")
print(f"  Total por Comisiones:    S/ {total_comisiones:.2f}")
print(f"  Sueldo Total a Recibir:  S/ {sueldo_total:.2f}")

```

5. Suponga que un individuo desea invertir su capital en un banco y desea saber cuánto dinero ganará después de un año si el banco paga a razón de 2% mensual.

>> IMPORTANTE: Se usará la formula:
>> - Valor Final = Valor Inicial x ( ( 1 x interes ) elevado al número del periodos )

```python
# Resolución de Ejercicio 5:
print("--- Simulación de 1 Año de inversión ---")

#Declaramos variables
MESES_AÑO: int = 12
capital_inicial: float = 0.0
interes_mensual: float = 0.0
capital_final: float = 0.0
interes_generado: float = 0.0


#Pedimos el capital inicial del mes 1
capital_inicial = float(input("\nIngrese el capital inicial a invertir: "))

#El interes lo pedimos al usuario y lo dividimos entre 100 para poder operarlo
interes_mensual = float(input("Ingrese la tasa de interes mensual: ")) / 100

print("\n--- Calculando interés en  12 meses---")

#Calculamos el capital final con la formula del interés compuesto:
# Valor Final = Valor Inicial x ( ( 1 x interes ) elevado al número del periodos )
capital_final = capital_inicial * ((1 + interes_mensual) ** MESES_AÑO)

#Calculamos unicamente el interes generado
interes_generado = capital_final - capital_inicial

#Imprimimos en pantalla los datos usando .2f para tener 2 decimales.
print("\n--- Proyección de Inversión a 1 Año ---")
print(f"  Capital Inicial: S/.{capital_inicial:.2f}")
print(f"  Tasa de Interés: 2.00% mensual")
print(f"  Período de Capitalización: {MESES_AÑO} meses")
print("---------------------------------------")
print(f"  Capital Final Acumulado: S/.{capital_final:.2f}")
print(f"  Interés Generado: S/.{interes_generado:.2f}")
```

6. Realice un programa que representen el algoritmo donde se ingrese la edad de un apersona para determinar aproximadamente cuántos meses, semanas, días y horas ha vivido una persona.

```python
# Resolución de Ejercicio 6:

print("--- Calculadora de Tiempo Vivido ---")

#Creamos constantes para calcular el tiempo en meses, semanas, dias y horas
MESES_POR_ANIO: int = 12
SEMANAS_POR_ANIO: int = 52 #Semanas aproximadas
DIAS_POR_ANIO: int = 365 #Dias aproximados
HORAS_POR_DIA: int = 24

# Inicialización de variables de el tiempo vivido
edad_anios: int = 0
meses_vividos: int = 0
semanas_vividas: int = 0
dias_vividos: int = 0
horas_vividas: int = 0

# Solicitamos la edad del usuario.
edad_anios = int(input("Ingrese su edad en años: "))

# Ahora vamos a calcular el equivalente multiplicamos la edad en año por cada valor
meses_vividos = edad_anios * MESES_POR_ANIO
semanas_vividas = edad_anios * SEMANAS_POR_ANIO
dias_vividos = edad_anios * DIAS_POR_ANIO

#Usamos los dias vividos para multiplicarlos por horas.
horas_vividas = dias_vividos * HORAS_POR_DIA

#Por ultimo imprimimos todo en pantalla
print("\n--- ¿Cuanto has vivido? ---")
print(f"  Tu edad es de: {edad_anios} años")
print(f"  Eso equivale a:")
print("----------------------------------------")
print(f"  {meses_vividos}   Meses vividos")
print(f"  {semanas_vividas}   Semanas vividas")
print(f"  {dias_vividos}   Días vividos")
print(f"  Y  {horas_vividas}  Horas vividas")
```

7. Un estacionamiento requiere determinar el cobro que debe aplicar a las personas que lo utilizan. Considere que el cobro es con base en las horas que lo disponen y que las fracciones de hora se toman como completas y realice un programa que permita determinar el cobro. El pago por hora es de S/5.00 y el pago por los minutos restantes de S/0.10 por minuto.

```python
# Resolución de Ejercicio 7:

print("----   Sistema de Cobro de Estacionamiento   ----")

# Declaramos constantes en las tarifas
TARIFA_POR_HORA: float = 5.00
TARIFA_POR_MINUTO: float = 0.10

# Inicialización de las variables
horas_estacionadas: int = 0
minutos_estacionados: int = 0
cobro_horas: float = 0.0
cobro_minutos: float = 0.0
cobro_total: float = 0.0

# Solicitamos al usuario ingresar laos datos de horas y minutos
horas_estacionadas = int(input("Ingrese la cantidad de horas completas: "))
minutos_estacionados = int(input("Ingrese la cantidad de minutos restantes: "))

# Para calcular el costo vamos a multiplicar las horas por 5 y los minutos por 0.10 y sumarlo
#Primero multiplicamos por separado
cobro_horas = horas_estacionadas * TARIFA_POR_HORA
cobro_minutos = minutos_estacionados * TARIFA_POR_MINUTO

#Ahora consolidamos sumando
cobro_total = cobro_horas + cobro_minutos

#Por ultimo mostramos en pantalla
print("\n----   Costo de Estacionamiento   ----")
print(f"  Costo de horas:   ({horas_estacionadas}):    S/ {cobro_horas:.2f}")
print(f"  Costo de minutos: ({minutos_estacionados}):    S/ {cobro_minutos:.2f}")
print("-----------------------------------------")
print(f"  Total a pagar:               S/ {cobro_total:.2f}")
```

8. Una empresa constructora vende terrenos con la forma A de la figura.
Obtener el área respectiva de un terreno de medidas ingresando el valor de los 3 lados.

>> IMPORTANTE: Usaremos:
>>   - Area del rectángulo: base x altura
>>   - Area del triángulo: (base x altura) / 2

```python
# Resolución de Ejercicio 8:

print("----   Sistema de cálculo de área del terrero   ----")

#Inicializamos las variables de los lados.
lado_a: float = 0.0
lado_b: float = 0.0
lado_c: float = 0.0


#Inicializamos las variables del triangulo.
alt_triangulo: float = 0.0
base_triangulo: float = 0.0
area_rectangulo: float = 0.0
area_triangulo: float = 0.0
area_total: float = 0.0


#Ahora solicitamos las medidas de los lados
print("\n----   Ingrese las medidas del terreno   ----")
lado_a = float(input("Ingresa las medidas del lado A:  "))
lado_b = float(input("Ingrese las medidas del lado B:  "))
lado_c = float(input("Ingrese las medidas del lado C:  "))

#Calculamos las medidas del triángulo
alt_triangulo = lado_a - lado_c #restamos lado A-C
base_triangulo = lado_b #La base del triángulo es la misma que la del rectángulo

#Calculamos las áreas
area_rectangulo = lado_b * lado_c #área del rectangulo base por lado
area_triangulo = (base_triangulo * alt_triangulo) / 2 #del triangulo base por altura entre 2
area_total = area_rectangulo + area_triangulo #Sumamos ambas áreas para el total

#Por ultimo mostramos los datos:
print("\n----   Ingrese las medidas del terreno   ----")
print(f"El área del rectángulo es:            {area_rectangulo}  m²")
print(f"El área del triángulo es:             {area_triangulo}  m²")
print(f"El área total del terreno es de:      {area_total}  m²")
```

9. Cuanto pagaría un cliente por una compra con descuento y cuanto sería el cambio que recibiría en un pago en efectivo. Ingresar el monto base, el porcentaje de descuento y el con cuánto dinero está pagando el cliente.

```python
# Resolución de Ejercicio 9:

print("----   Sistema de Caja   ----")

# Inicializamos las variables
monto_base: float = 0.0
porcentaje_descuento: float = 0.0
monto_pago: float = 0.0
monto_descuento: float = 0.0
total_pagar: float = 0.0
cambio_cliente: float = 0.0

# Pedimnos el monto base de la compra
monto_base = float(input("Ingrese el monto de la compra: S/ "))

# Pedimos el porcentaje de descuento
porcentaje_descuento = float(input("Ingrese el porcentaje de descuento: "))

monto_pago = float(input("Ingrese el la cantidad de dinero en efectivo con el que pagará: S/ "))

# Ahora para saber el comto de descuento usamos el monto base
# Tenemos que dividir el porcentaje de descuento entre 100 para poder calcularlo.
monto_descuento = monto_base * (porcentaje_descuento / 100)

# El total a pagar será el monto base menos el mento del descuento
total_pagar = monto_base - monto_descuento

# Para calcular el cambio sencillamente restamos lo que tiene que pagar al efectivo del cliente
cambio_cliente = monto_pago - total_pagar

# Mostramos los resultados finales en pantalla y con dos decimales (.2f)
print("\n----   Detalle de la Compra   ----")
print(f"  Monto base:               S/ {monto_base:.2f}")
print(f"  Descuento de ({porcentaje_descuento}%):     -S/ {monto_descuento:.2f}")
print("-----------------------------------------")
print(f"  Total a pagar:            S/ {total_pagar:.2f}")
print(f"  Pago del cliente:         S/ {monto_pago:.2f}")
print(f"  Cambio a devolver:        S/ {cambio_cliente:.2f}")
```
