1.  Calcular el total que una persona debe pagar en una llantera, si el precio de cada llanta es de S/150 si se compran menos de 5 llantas y de S/120 si se compran 5 o más.

```python
# Resolución de Ejercicio 1:

print("--- Calculando precio de llantas---")

#Pedimos la cantidad de llantas
cantidad_llantas: int = 0

#Inicializamos una variable del precio y el total
precio_unitario: float = 0.0
total_pagar: float = 0.0


#Pedimos la cantidad de llantas
cantidad_llantas = int(input("Ingrese la cantidad de llantas a comprar: "))

#Con una condicional identificamos el precio según la cantidad de llantas
if cantidad_llantas < 5:
    precio_unitario = 150.0
# Si la condición anterior es falsa (es decir, la cantidad es 5 o mayor), el precio es S/ 120.
else:
    precio_unitario = 120.0

#Por ultimo calculamos el total a pagar multiplicando el precio por la cantidad de llantas
total_pagar: float = cantidad_llantas * precio_unitario

# 5. Salida de datos estructurada
print("\n--- Precio total de la compra: ---")
print(f"  Cantidad solicitada: {cantidad_llantas} llanta(s)")
print(f"  Precio unitario aplicado: S/ {precio_unitario:.2f}")
print("--------------------------------")
print(f"  Total a pagar: S/ {total_pagar:.2f}")
```

2. Cómo se podría determinar ¿Cuál de los 3 números brindados por usuario es el menor y mayor de todos ellos?

```python
# Resolución de Ejercicio 2:

print("----   Análisis de Numéros   ----")

#Declaramos variables
numero_uno: float = 0.0
numero_dos: float = 0.0
numero_tres: float = 0.0

numero_mayor: float = 0.0
numero_menor: float = 0.0


#Solicitamos los 3 números
numero_uno = float(input("Ingrese el primer número: "))
numero_dos = float(input("Ingrese el segundo número: "))
numero_tres = float(input("Ingrese el tercer número: "))

# Usaremos el número UNO como referencia ya que es el primer número ingresado.
numero_mayor: float = numero_uno
numero_menor: float = numero_uno

#Ahora comprobamos con una condicional si el número dos es mayor al número mayor (que es el uno)
if numero_dos > numero_mayor:
    numero_mayor = numero_dos
#Aqui no necesitamos un else ni elif
#Si la primera condición no se cumple se ignora y pasa a comprobar la siguiente
#Ahora verificamos si el numero 3 es mayor.
if numero_tres > numero_mayor:
    numero_mayor = numero_tres

#Repetimos el mismo procedimiento para poder comparar cual es menor.
if numero_dos < numero_menor:
    numero_menor = numero_dos

if numero_tres < numero_menor:
    numero_menor = numero_tres

#Imprimimos en pantalla
print("\n----   Resultados del Análisis   ----")
print(f"  Los números ingresados fueron: \n {numero_uno}, \n {numero_dos} y \n {numero_tres}")
print("---------------------------------------------")
print(f"  El número MAYOR es: {numero_mayor}")
print(f"  El número MENOR es: {numero_menor}")
```

4. En una escuela la colegiatura de los alumnos se determina según el número de materias que cursan (Solicitar al usuario). El costo de todas las materias es el mismo S/100. Se ha establecido un programa para estimular a los alumnos, el cual consiste en lo siguiente: si el promedio obtenido por un alumno en el último periodo es mayor o igual que 18, se le hará un descuento del 50% sobre la mitad de los cursos; si el promedio obtenido es menor que 18 deberá pagar la colegiatura completa.
```python
# Resolución de Ejercicio 4:

print("--- Cálculo de Pago de Colegiatura ---")

#Primero declaramos las variables de entrada
COSTO_MATERIA: int = 100
PORCEN_DESCUENTO_MATERIAS: float = 0.5
materias: int = 0
promedio: float = 0.0

#Declaramos las variables de salida
colegiatura: float = 0.0
descuento: float = 0.0
total_pagar: float = 0.0


#Solicitamos al usuario los datos de cantidad de materias y promedio
materias = int(input("Ingrese el número de materias que cursa: "))
promedio = float(input("Ingrese el promedio del último período: "))

print("\n--- Calculando Costo ---")

#Calculamos el costo de la colegiatura  multiplicandolo por la cantidad
colegiatura = materias * COSTO_MATERIA

#Ahora si el promedio es mayor o igual a se aplica un 50% sobre la mitad del costo
if promedio >= 18:
    #Asi que dividimos entre 2 la colegiatura y al resto lo multiplicamos por el descuento
    descuento = (colegiatura / 2) * PORCEN_DESCUENTO_MATERIAS
    total_pagar = colegiatura - descuento

    #Imprimimos en pantalla el descuento
    print("\n    El alumno recibe descuento.")
    print(f"\n    escuento: S/ {descuento:.2f}")
else: #Sino, el pago es el total de la colegiatura
    total_pagar = colegiatura
    #Imprimimos en pantalla
    print("\n    El alumno no recibe descuento.")

print(f"\n    Total a pagar: S/ {total_pagar:.2f}")

print("\n--- PROGRAMA TERMINADO ---")
```

5. Determinar mediante el ingreso de los 3 lados de un triángulo, si este es Equilátero (Mostrar el perímetro del), Isósceles (Indicar que lados son Iguales) o Escaleno (triángulo mostrar el área del triángulo).

```python
# Resolución de Ejercicio 5:
#Necesitamos importar math para poder usar raiz cuadrada
#Esto es necesario porque el area de un triangulo sin saber su altura necesita la raiz
import math

#Declaramos las variables
perimetro: float = 0.0
semi_peri: float = 0.0
area: float = 0.0
lado_a: float = 0.0
lado_b: float = 0.0
lado_c: float = 0.0

print("--- Determinando tipo de triángulo ---")

#Solicitamos los datos de entrada de los lados
lado_a = float(input("\nIngrese el valor del lado A: "))
lado_b = float(input("Ingrese el valor del lado B: "))
lado_c = float(input("Ingrese el valor del lado C: "))

#En la primer condicional verificamos con una triple validación si los 3 lados son iguales
if lado_a  == lado_b and lado_a == lado_c and lado_b == lado_c :
  perimetro  = lado_a + lado_b + lado_c #Calculamos el perimetro sumando los lados
  #Mostramos en pantalla el resultado
  print("\nLos lados del triangulo son iguales")
  print("\n   El triangulo ES EQUILATERO")
  print(f"\n El perímetro del triángulo es: {perimetro}")

#Sino verificamos con tripli condicional usando "or" si 2 lados son iguales
elif lado_a == lado_b or lado_a == lado_c or lado_b == lado_c:
  #Imprimimos en pantalla
  print("\nSolo dos lados del triangulo son iguales")
  print("\n   El triangulo ES ISÓSCELES")

  #Con un conjunto de condicionales verificamos cual de los lados es igual
  if lado_a == lado_b:
      print(f"\n   Los lados iguales son: {lado_a} y {lado_b}")

  elif lado_a == lado_c:

    print(f"\n   Los lados iguales son: {lado_a} y {lado_c}")

  else:
      print(f"\n   Los lados iguales son: {lado_b} y {lado_c}")

else: #Sino es ninguno de las anteriores es escaleno
  print("\nNingun lado del triangulo es igual")
  print("\n   El triangulo ES ESCALENO")

  #Como no sabemos la altura necesiamos el perímetro para sacar el área.
  perimetro = lado_a + lado_b + lado_c
  print(f"\n   El perímetro del triángulo es: {perimetro}")

  #Ahora usamos el perimetro para sacar el area
  #Sacamos el semi perímetro
  semi_peri: float = perimetro / 2

  #Ahora usamos la formula para sacar el area sin altura:
  area: float = math.sqrt(semi_peri * (semi_peri - lado_a) * (semi_peri - lado_b) * (semi_peri - lado_c))
  print(f"\n   El área del triángulo es: {area}")

print("--- Programa terminado ---")


```

6. Elabore un programa que lea la hora (HH MM SS) y muestre por
pantalla la hora N segundos después.

```python
# Resolución de Ejercicio 6:

print("----   Sistema de Cálculo de Tiempo   ----")

# Inicialización de variables
horas: int = 0
minutos: int = 0
segundos: int = 0
n_segundos: int = 0

# Ahora solicitamos la hora inicial
print("       Ingrese la hora actual:")
horas = int(input("         Horas (0-23): "))
minutos = int(input("         Minutos (0-59): "))
segundos = int(input("          Segundos (0-59): "))
# Luego los segundos a sumar
n_segundos = int(input("\nIngrese la cantidad de segundos a sumar (N): "))

# Una ves tenemos los segundos a añadir los acumulamos a la variable segundos
segundos = segundos + n_segundos

# Ahora verificamos si los segundos superan 60, de hacerlo aumentamos 1 min
if segundos >= 60:
    # Si es mayor a 60 actualizamos minutos sumando la división entera de segundos
    minutos = minutos + (segundos // 60)
    # Los actualizamos los segundos añadiendo el residuo
    segundos = segundos % 60

# Repetimos el mismo procedimiento
if minutos >= 60:
    horas = horas + (minutos // 60) #Actualizamos los minutos
    minutos = minutos % 60 # El residuo se queda como los minutos finales

if horas >= 24:
    # Si sobrepasa las 24 horas, el residuo nos da la hora del día siguiente
    horas = horas % 24

# Imprimimos los resultados
print("\n----   Resultado del Salto Temporal   ----")
# Utilizamos ":02d" para forzar el relleno con ceros a la izquierda
print(f"  Nueva hora calculada: {horas:02d}:{minutos:02d}:{segundos:02d}")
```

7. Calcular el pago de una llamada, el cobro es por el tiempo que ésta dura, de tal forma que los primeros cinco minutos cuestan S/ 1.00 c/u, los siguientes tres, S/0.80 c/u, los siguientes dos minutos, S/0.70 c/u,3 y a partir del décimo minuto, S/0.50 c/u. Además, se carga un impuesto de 3 % cuando es domingo, y 5% si es domingo de turno nocturno. Realice un algoritmo para determinar cuánto debe pagar por cada concepto una persona que realiza una llamada.

```python
# Resolución de Ejercicio 7:

print("----   Sistema de Facturación   ----")

# Creamos la variables

COSTO_5: float = 1.00
COSTO_8: float = 0.80
COSTO_10: float = 0.70
COSTO_MAYOR_10: float = 0.50

duracion_minutos: int = 0
dia_semana: str = ""
turno_llamada: str = ""

costo_base: float = 0.0
tasa_impuesto: float = 0.0
monto_impuesto: float = 0.0
total_pagar: float = 0.0

# Pedimos los minutos a usar
duracion_minutos = int(input("Ingrese la duración de la llamada en minutos: "))

# Pedimos el dia para saber si es domingo
dia_semana = input("¿Qué día de la semana se realizó la llamada? ").strip().lower()

# Ahora con condicionales sumaremos escalonadamente el costo en minutos
# Si dura 5 aplicamos la tarifa
if duracion_minutos <= 5:
    costo_base = duracion_minutos * COSTO_5

#Sino sumamos los minutos anteriores y aplicamos la siguiente tarifas
elif duracion_minutos <= 8:
    costo_base = ( COSTO_5 * 5 ) + ((duracion_minutos - 5) * COSTO_8)

#Repetimos el proceso sumando lo consumido en los minutos anteriores
elif duracion_minutos <= 10:
    costo_base = ( COSTO_5 * 5 ) + ( COSTO_8 * 3 )  + ((duracion_minutos - 8) * COSTO_10)

#Si no se cumple ninguna solución se suman todas las tarifas y al restante se le aplica la tarifa final
else:
    costo_base = ( COSTO_5 * 5 ) + ( COSTO_8 * 3 ) + ( COSTO_10 * 2 ) + ((duracion_minutos - 10) * COSTO_MAYOR_10)

# Evaluamos si el dia es domingo
if dia_semana == "domingo":
   # Usamos stip y lower para quitar espacios en blanco y que sea todo minúscula
   # Si es domingo preguntamos el turno
    turno_llamada = input("¿La llamada fue en turno nocturno? (s/n): ").strip().lower()

    if turno_llamada == 's':
        tasa_impuesto = 0.05  # 5% si es domingo nocturno
    else:
        tasa_impuesto = 0.03  # 3% si es domingo diurno/matutino
else:
    tasa_impuesto = 0.00  # 0% para cualquier otro día

# Finalmente calculamos los montos totales
monto_impuesto = costo_base * tasa_impuesto
total_pagar = costo_base + monto_impuesto

# Imprimimos en pantalla
print("\n----   Desglose de Facturación   ----")
print(f"  Duración total:        {duracion_minutos} minuto(s)")
print(f"  Costo base por tiempo: S/ {costo_base:.2f}")
print("-----------------------------------------")
print(f"  Impuesto aplicado:     {(tasa_impuesto * 100):.1f}%")
print(f"  Cargo por impuesto:    S/ {monto_impuesto:.2f}")
print("-----------------------------------------")
print(f"  Total a pagar:         S/ {total_pagar:.2f}")
```

8. Solicite a usuario 3 números enteros (validar que sean diferentes) y en que orden lo desean ver, para luego, mostrarlos a usuario, pero ordenados de manera ascendente o descendente según los que indique el usuario

```python
# Resolución de Ejercicio 8:

print("----   Sistema de Orden Numérico   ----")

#Inciamos las variables
numero_uno: int = 0
numero_dos: int =  0
numero_tres: int = 0
orden_deseado: str = ""

menor: int = 0
medio: int = 0
mayor: int = 0

# Pedimos los datos
numero_uno: int = int(input("\nIngrese el primer número entero: "))
numero_dos: int = int(input("Ingrese el segundo número entero: "))
numero_tres: int = int(input("Ingrese el tercer número entero: "))

# Verificamos que los números sean diferentes
if numero_uno == numero_dos or numero_uno == numero_tres or numero_dos == numero_tres:
    print("\n  [ERROR] Los tres números debens ser diferentes.")
else:
    #Preguntamos el orden con strip y upper para que la respues sea siempre mayuscula y no tenga espacios
    orden_deseado = input("\n¿En qué orden desea verlos? (Ingrese 'A' para Ascendente o 'D' para Descendente): ").strip().upper()

    # Ahora verificamos el menor

    # Verificamos si el primero es el menor
    if numero_uno < numero_dos and numero_uno < numero_tres:
        menor = numero_uno
        #Si lo es verificamos si el segundo es el del medio
        if numero_dos < numero_tres:
            medio = numero_dos
            mayor = numero_tres
        else: #Si no lo es entonces el número 3 es el del medio
            medio = numero_tres
            mayor = numero_dos

    # Si el primer numero no es menor repetirmos el mismo proceso pero verificando si el segundo número es menor
    elif numero_dos < numero_uno and numero_dos < numero_tres:
        menor = numero_dos
        if numero_uno < numero_tres:
            medio = numero_uno
            mayor = numero_tres
        else:
            medio = numero_tres
            mayor = numero_uno

    # Si ni el segundo ni el primero son menores, por lo tanto el tercero es el menor
    else:
        menor = numero_tres

        #Solo debemos verificar cual es el del medio
        if numero_uno < numero_dos:
            medio = numero_uno
            mayor = numero_dos
        else:
            medio = numero_dos
            mayor = numero_uno

    # Imprimimos en pantalla
    print("\n----   Resultados del Ordenamiento   ----")
    if orden_deseado == 'A':
        print(f"  Orden Ascendente:  {menor}, {medio}, {mayor}")
    elif orden_deseado == 'D':
        print(f"  Orden Descendente: {mayor}, {medio}, {menor}")
    else:
        print("  [Error de Formato] Debe ingresar 'A' o 'D'.")

    print("\n----   FIN DEL PROGRAMA   ----")
```

#
9. Permita calcular el total a pagar por la compra de N camisas. Si se compran entre 1 a 4 camisas se aplica un descuento del 12.5%, si se compra una cantidad comprendida entre 5 y 8 camisas se aplica un descuento del 20% y si se compran cantidades mayores, se aplica un descuento del 31.5% sobre el total de la compra. Debe imprimirse en pantalla la compra final sin descuento, monto del descuento y la compra con descuento.

```python
# Resolución de Ejercicio 9:
print("--- Cálculo de pago e camisas---")

#Pedimos la cantidad de camisas compradas
cantidad_camisas: int = int(input("Ingrese la cantidad de camisas a comprar: "))

#Declaramos las variables
precio_unitario: float = 0.0
porcentaje_descuento: float = 0.0
total: float = 0.0
total_descuento: float = 0.0
total_pagar: float = 0.0

#Nos aseguramos que es mayor a 0 y si no lo es se lanza un error y se acaba el programa
if cantidad_camisas <= 0:
    print("  [ERROR] Ingrese una cantidad mayor a 0.")
    print("  Fin del programa.")

#Si es mayor a 0 entonces comenzamos al validación
else:
    #Pedimos ingresar el precio como flotante porque este puede ser decimal.
    precio_unitario = float(input("Ingrese el precio por camisa: "))

    # Inicialización una variable en 0 para poder asignarle el descuento
    porcentaje_descuento = 0.0

    #Con uan condicional verificamos en que rango de cantidad de camisas se encuentra.
    #Se usan decimales ya que el procentaje se representa con su división entre 100
    if 1 <= cantidad_camisas <= 4:
        porcentaje_descuento = 0.125 # 12.5 / 100
    elif 5 <= cantidad_camisas <= 8:
        porcentaje_descuento = 0.20 # 20.0 / 100
    else:
        #Si no es ninguno de los anteriores entonces compro más de 9 camisas.
        #Corresponde el maximo porcentaje de descuento
        porcentaje_descuento = 0.315 # 31.5 / 100

    #Ahora calculamos el total sin el descuento
    total = cantidad_camisas * precio_unitario

    #Ahora calculamos el total del descuento
    total_descuento = total * porcentaje_descuento

    #Por último calculamos el precio total a pagar con el descuento
    total_pagar = total - total_descuento

    #Imprimimos en Pantalla usando .2f para tener únicamente 2 decimales.
    print("\n--- Montos totales ---")
    print(f"  Cantidad de camisas compradas: {cantidad_camisas} camisa(s)")
    print(f"  Porcentaje de descuento: {porcentaje_descuento * 100}%")
    print("---------------------------------------")
    print(f"  Total sin descuento:         $  {total:.2f}")
    print(f"  Total del descuento:        -$  {total_descuento:.2f}")
    print(f"  Total neto a pagar:          $  {total_pagar:.2f}")
```
